```python
from pyspark.sql.functions import sum as _sum, count

cat = "notebook_catalog.notebook_schema"

gold = (spark.table(f"{cat}.sales_silver")
    .groupBy("product")
    .agg(_sum("quantity").alias("total_quantity"),
         _sum("total_amount").alias("total_sales"),
         count("*").alias("num_orders")))

gold.write.mode("overwrite").saveAsTable(f"{cat}.sales_gold")
display(gold)
```
